# Taking max or average for uncertainties

When converting uncertainties to log space, linmix requires just one value -- take the max or average?

Way 1:
- Take the average 

Way 2:
- Take the max. 


Results:
- I choose way 2.
- Way 2 is more conservative in terms of the uncertainties. 

```python
## Example with very asymmetric uncertainty
y = 8.07e-12
dy_l = 8.06e-12
dy_u = 3e-14
x = np.log10(y)

## Way 1
# Convert to log
dx = 0.5*(np.log10(y+ dy_u) - np.log10(y - dy_l))
print(dx)
# Converting back
yerr_l = y - 10**(x- dx)
yerr_u = 10**(x + dx) - y
print("dy_l:", yerr_l)
print("dy_u:", yerr_u)

print()
## Way 2
dx = np.max([np.log10(y+ dy_u) -  np.log10(y), np.log10(y) - np.log10(y -  dy_l) ])
print(dx)
# Converting back
yerr_l = y - 10**(x- dx)
yerr_u = 10**(x + dx) - y
print("dy_l:", yerr_l)
print("dy_u:", yerr_u)


## Using simple propagation of uncertainties formula (only valid in low SNR case)
print()
dx_l = 1/np.log(10) * dy_l/y
dx_u = 1/np.log(10) * dy_u/y
print(np.log10(y) - np.log10(y -  dy_l) )
print(dx_l)
print(np.log10(y+ dy_u) -  np.log10(y))
print(dx_u)
```

# Investigate adding systematic 10% uncertainty in linear vs log space. 

Way 1:
- Add 10% uncertainty in linear space first, then convert to log space. 

Way 2: 
- Add 10% uncertainty in log space.


Results:
- I opt to use way 1

```python
# Systematic uncertainty when added in log space; dy_u = dy_l = 0.1*y
# dlogy_u = log10(y + dy_u) - log10(y) = log10(1.1)
# dlogy_l = log10(y) - log10(y - dy_l) = log10(1/0.9)
print(np.log10(1.1))
print(np.log10(1/0.9))
```

## High uncertainty

```python
## Example with high uncertainty
y = 8.07e-12
dy_l = 2.2e-12
dy_u = 2.2e-12

x = np.log10(y)

## Way 1
dy_l_new = np.sqrt(dy_l**2 + (0.1*y)**2)
dy_u_new = np.sqrt(dy_u**2 + (0.1*y)**2)
#print(dy_l_new, dy_u_new)
# Convert to log
dx_u = np.log10(y+ dy_u_new) - np.log10(y)
dx_l = np.log10(y) - np.log10(y - dy_l_new)
print(dx_l, dx_u)
print(y - 10**(x - dx_l), 10**(x + dx_u) - y)

print()
## Way 2
dx_sys_u =  np.log10(1.1)
dx_sys_l = np.log10(1/0.9)
dx_fit_u = np.log10(y+ dy_u) - np.log10(y)
dx_fit_l = np.log10(y) - np.log10(y - dy_l)
dx_l = np.sqrt(dx_sys_l**2 + dx_fit_l**2)
dx_u = np.sqrt(dx_sys_u**2 + dx_fit_u**2)
print(dx_l, dx_u)
print(y - 10**(x - dx_l), 10**(x + dx_u) - y)
```
