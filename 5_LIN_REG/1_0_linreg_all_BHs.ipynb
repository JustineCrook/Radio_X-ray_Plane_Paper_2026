```python
# Test ... we 1e32, ~4.5-4.9e26
Lr_med = 4.54e+28 #erg/s
Lx_med = 7.51e+35 #erg/s

beta = 0.57
alpha = 0.22

Lx = 1e32
Lr = Lr_med * 10**alpha * (Lx/Lx_med)**beta
print(f"Predicted Lr for Lx={Lx:.2e} erg/s: {Lr:.2e} erg/s")

# Result should be around 5e25
Lx = 2e30
Lr = Lr_med * 10**alpha * (Lx/Lx_med)**beta
print(f"Predicted Lr for Lx={Lx:.2e} erg/s: {Lr:.2e} erg/s")
```

### Test 2 runs

```python
run_linmix(N_runs=2, interp=True, type_source="BH", K=3, seed=5, parallel=False, verbose=True, testing=True)
```

```python
""" 
d_corr for this source:  [0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671 0.01090671]
Lx before any transformation:  [1.99635482e+37 3.43202106e+37 1.88782598e+36 2.30870315e+35 2.96021971e+34]
Lx after transformation: [ 1.4245978   1.65991001  0.40032202 -0.51227184 -1.40431599]
Lx after dcorr: [ 1.43550451  1.67081672  0.41122873 -0.50136513 -1.39340928]
Lr before any transformation:  [1.01553095e+29 2.11778972e+29 2.54179243e+28 2.39428100e+27 1.98658609e+27]
Lr after transformation: [ 0.34963731  0.66882698 -0.25191577 -1.27788073 -1.35894846]
Lr after dcorr: [ 0.36054402  0.6797337  -0.24100906 -1.26697402 -1.34804175]
"""
```

```python
Lr_med = 4.54e+28 #erg/s
Lx_med = 7.51e+35 #erg/s

D= 2.2

# Sampled distance at second iteration
Di = D* 10**(0.01090671/2)
print("Di: ", Di)   # higher than D, so we expect Lr and Lx to increase after dcorr, which is what we see in the results above


# Lx transformation
Lx = np.array([1.99635482e+37, 3.43202106e+37, 1.88782598e+36, 2.30870315e+35, 2.96021971e+34])
lx = np.log10(Lx/Lx_med)
print("lx after transformation: ", lx[:5]) # same as above
# After dcorr -- alt
Lx_new = Lx * (Di/D)**2
print("Lx after dcorr: ", Lx_new[:5]) 
print("lx after dcorr: ", np.log10(Lx_new/Lx_med)[:5]) # same as above

print()
Lr = np.array([1.01553095e+29, 2.11778972e+29, 2.54179243e+28, 2.39428100e+27, 1.98658609e+27])
lr = np.log10(Lr/Lr_med)
print("lr after transformation: ", lr[:5]) # same as above
# After dcorr -- alt
Lr_new = Lr * (Di/D)**2
print("Lr after dcorr: ", Lr_new[:5]) 
print("lr after dcorr: ", np.log10(Lr_new/Lr_med)[:5]) # same as above
```
